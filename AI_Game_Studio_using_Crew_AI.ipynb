Create Tools.py

```python
%%writefile tools.py

"""
============================================================
tools.py

Defines all tools used by the AI Game Studio.

Compatible with:
- CrewAI v1.15.5
- CrewAI Tools
============================================================
"""

# ============================================================
# Imports
# ============================================================

import io
import os
import traceback
import contextlib
from typing import Type

from pydantic import BaseModel, Field

from crewai.tools import BaseTool

from crewai_tools import (
    SerperDevTool,
    FileWriterTool,
    FileReadTool,
    DirectoryReadTool,
)

# ============================================================
# Create Output Directory
# ============================================================

# Create outputs folder if it doesn't exist.
# All generated files will be stored here.
os.makedirs("outputs", exist_ok=True)

# ============================================================
# File Tools
# ============================================================

# Tool for saving files generated by AI agents
file_writer = FileWriterTool()

# Tool for reading a specific file
file_reader = FileReadTool()

# Tool for reading files inside the outputs folder
directory_reader = DirectoryReadTool(
    directory="outputs"
)

# ============================================================
# Internet Search Tool (Optional)
# ============================================================

# Loads Serper only if SERPER_API_KEY exists.
# If no API key is available, the project will
# continue to work without internet search.

try:
    search_tool = SerperDevTool()
    print("✅ Serper Search Tool Loaded")

except Exception:
    search_tool = None
    print("⚠️ Serper API Key not found. Search Tool Disabled.")

# ============================================================
# Python Code Executor Tool
# ============================================================

class CodeExecutorInput(BaseModel):
    """
    Input schema for the Python Code Executor.
    """

    code: str = Field(
        ...,
        description="Python code to execute."
    )


class PythonCodeExecutorTool(BaseTool):
    """
    Executes Python code and returns runtime output.

    Used by the Code Reviewer Agent to validate
    generated game code.
    """

    # Tool name visible to CrewAI
    name: str = "Python Code Executor"

    # Tool description
    description: str = (
        "Executes Python code and returns "
        "runtime output or execution errors."
    )

    # Input schema
    args_schema: Type[BaseModel] = CodeExecutorInput

    # --------------------------------------------------------
    # Execute Python Code
    # --------------------------------------------------------

    def _run(self, code: str) -> str:

        output = io.StringIO()

        try:

            # Capture all print() output
            with contextlib.redirect_stdout(output):

                # Execute inside an isolated namespace.
                # Intended only for validating generated code.
                # For production systems, use a secure sandbox.
                exec(
                    code,
                    {"__builtins__": __builtins__},
                    {}
                )

            result = output.getvalue().strip()

            if result:
                return (
                    "✅ Code executed successfully.\n\n"
                    + result
                )

            return "✅ Code executed successfully."

        except Exception as e:

            return (
                "❌ Execution Failed\n\n"
                f"Error: {str(e)}\n\n"
                f"{traceback.format_exc()}"
            )


# ============================================================
# Initialize Custom Tool
# ============================================================

code_executor = PythonCodeExecutorTool()

# ============================================================
# Confirmation Message
# ============================================================

print("✅ tools.py loaded successfully")
```

```python
%%writefile agents.py

"""
============================================================
agents.py

Defines all AI agents used in the AI Game Studio.

Compatible with:
- CrewAI v1.15.5
- OpenRouter API
============================================================
"""

# ============================================================
# Imports
# ============================================================

import os

from crewai import Agent, LLM

from tools import (
    search_tool,
    file_writer,
    file_reader,
    directory_reader,
    code_executor
)


# ============================================================
# Configure OpenRouter LLM
# ============================================================


# Create shared LLM using OpenRouter

llm = LLM(
    model="ollama/llama3.2:3b",  # or 7b
    base_url="http://127.0.0.1:11434"
)



# ============================================================
# Tool Lists
# ============================================================

designer_tools = []

if search_tool:
    designer_tools.append(search_tool)


developer_tools = [file_writer]

if search_tool:
    developer_tools.insert(0, search_tool)


reviewer_tools = [
    directory_reader,
    file_reader,
    code_executor
]


tester_tools = [
    directory_reader,
    file_reader
]


writer_tools = [
    file_writer
]


# ============================================================
# Project Manager Agent
# ============================================================

project_manager = Agent(

    role="Project Manager",

    goal="""
Plan the entire game development workflow,
coordinate all AI agents,
track progress,
and ensure successful project completion.
""",

    backstory="""
You are an experienced Game Producer who has
managed multiple successful indie game projects.

You assign work,
manage deadlines,
monitor progress,
and ensure quality delivery.
""",

    llm=llm,

    verbose=True,

    memory=True,

    allow_delegation=True,

    tools=[file_writer]
)


# ============================================================
# Game Designer Agent
# ============================================================

designer = Agent(

    role="Game Designer",

    goal="""
Design engaging gameplay,
interesting mechanics,
balanced difficulty,
and enjoyable player experience.
""",

    backstory="""
You are a professional indie Game Designer.

You design:

• Story
• Characters
• Gameplay
• Levels
• Progression
• Rewards
""",

    llm=llm,

    verbose=True,

    memory=True,

    allow_delegation=False,

    tools=designer_tools
)


# ============================================================
# Python Game Developer
# ============================================================

developer = Agent(

    role="Python Game Developer",

    goal="""
Develop a complete,
clean,
well-structured,
production-quality Pygame game.
""",

    backstory="""
You are a senior Python Game Developer
specializing in Pygame.

You write modular,
object-oriented,
well-commented,
efficient code.
""",

    llm=llm,

    verbose=True,

    memory=True,

    allow_delegation=False,

    tools=developer_tools
)


# ============================================================
# Code Reviewer
# ============================================================

reviewer = Agent(

    role="Code Reviewer",

    goal="""
Review generated code,
identify bugs,
verify syntax,
improve readability,
and ensure PEP8 compliance.
""",

    backstory="""
You are a Senior Software Engineer responsible
for reviewing production-ready Python projects
before release.
""",

    llm=llm,

    verbose=True,

    memory=True,

    allow_delegation=False,

    tools=reviewer_tools
)


# ============================================================
# QA Tester
# ============================================================

tester = Agent(

    role="QA Tester",

    goal="""
Test gameplay,
verify functionality,
find edge cases,
and prepare a professional QA report.
""",

    backstory="""
You are an experienced QA Engineer
specializing in game testing.

You validate gameplay,
performance,
stability,
and user experience.
""",

    llm=llm,

    verbose=True,

    memory=True,

    allow_delegation=False,

    tools=tester_tools
)


# ============================================================
# Documentation Writer
# ============================================================

writer = Agent(

    role="Technical Writer",

    goal="""
Create professional documentation,
README,
installation guide,
controls,
and project summary.
""",

    backstory="""
You are an expert technical writer.

You create professional,
GitHub-ready,
well-structured Markdown documentation.
""",

    llm=llm,

    verbose=True,

    memory=True,

    allow_delegation=False,

    tools=writer_tools
)


print("✅ agents.py loaded successfully with OpenRouter")
```

Create crew.py

```python
%%writefile crew.py

"""
============================================================
crew.py

Creates the AI Game Studio Crew.

This file combines:
1. Agents
2. Tasks
3. Workflow

Compatible with:
- CrewAI v1.15.5
- OpenRouter API
============================================================
"""


# ============================================================
# Imports
# ============================================================

from crewai import Crew, Process


# Import all AI agents

from agents import (
    project_manager,
    designer,
    developer,
    reviewer,
    tester,
    writer
)


# Import all tasks

from tasks import (
    planning_task,
    design_task,
    development_task,
    review_task,
    testing_task,
    documentation_task
)



# ============================================================
# Create AI Game Studio Crew
# ============================================================

game_builder_crew = Crew(

    # --------------------------------------------------------
    # Agents participating in workflow
    # --------------------------------------------------------

    agents=[
        project_manager,
        designer,
        developer,
        reviewer,
        tester,
        writer
    ],


    # --------------------------------------------------------
    # Tasks executed in order
    # --------------------------------------------------------

    tasks=[
        planning_task,
        design_task,
        development_task,
        review_task,
        testing_task,
        documentation_task
    ],


    # --------------------------------------------------------
    # Workflow Configuration
    # --------------------------------------------------------

    # Execute tasks sequentially
    process=Process.sequential,


    # Show detailed execution logs
    verbose=True,


    # Enable shared memory
    memory=True,


    # Cache repeated responses
    cache=True,


    # Disable automatic planner
    # Helps reduce API calls
    planning=False,


    # OpenRouter rate limit protection
    max_rpm=5
)



# ============================================================
# Confirmation Message
# ============================================================

print("✅ crew.py loaded successfully with OpenRouter")
```

Create main.py

```python
%%writefile main.py

"""
============================================================
AI Game Studio

Entry point for the project.

Workflow
--------
1. Load configuration
2. Import Crew
3. Create output folder
4. Get game idea
5. Execute CrewAI workflow
6. Display results
7. List generated files

Compatible with:
- CrewAI v1.15.5
- OpenRouter API
- Google Colab
- Jupyter Notebook
- Local Python
============================================================
"""

# ============================================================
# Imports
# ============================================================

import os
import asyncio
import traceback
import nest_asyncio

# Allow nested event loops (required in Google Colab)
nest_asyncio.apply()

# ============================================================
# Load Configuration
# ============================================================

# Loads OpenRouter API Key and optional Serper API Key
import config

# ============================================================
# Import Crew
# ============================================================

from crew import game_builder_crew

# ============================================================
# Create Output Directory
# ============================================================

os.makedirs("outputs", exist_ok=True)

print("✅ AI Game Studio Initialized")

# ============================================================
# Main Function
# ============================================================

async def main():

    print("=" * 60)
    print("🎮 AI GAME STUDIO")
    print("=" * 60)

    game_idea = input("\n🎮 Enter your game idea: ").strip()

    if not game_idea:
        print("❌ Game idea cannot be empty.")
        return

    print("\n🚀 Starting AI Game Studio...")
    print("Please wait while the AI team builds your game...\n")

    try:

        result = await game_builder_crew.kickoff_async(
            inputs={
                "game_idea": game_idea
            }
        )

    except Exception as e:

        print("\n❌ Crew Execution Failed!\n")
        print(f"Error: {e}")
        traceback.print_exc()
        return

    print("\n" + "=" * 60)
    print("🎉 PROJECT COMPLETED!")
    print("=" * 60)

    print("\nFinal Project Summary:\n")
    print(result)

    print("\nGenerated Files\n")

    generated = False

    expected_files = [
        "game.py",
        "README.md",
        "code_review.md",
        "test_report.md"
    ]

    for file in expected_files:

        path = os.path.join("outputs", file)

        if os.path.exists(path):
            generated = True
            print(f"✅ {path}")
        else:
            print(f"❌ Missing: {path}")

    if not generated:
        print("\n⚠️ No output files were generated.")

    print("\n🏁 AI Game Studio Finished Successfully!")


# ============================================================
# Run Application
# ============================================================

if __name__ == "__main__":

    try:
        asyncio.run(main())

    except RuntimeError:
        loop = asyncio.get_event_loop()
        loop.run_until_complete(main())
```
